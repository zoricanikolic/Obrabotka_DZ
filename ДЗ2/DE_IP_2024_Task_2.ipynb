```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

# Наложение и удаление шума

```python
image = cv2.imread('sar_1.jpg')
image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) 
```

```python
plt.imshow(image_gray, cmap="gray")
```

```python
# Gaussian noise
mean = 0
stddev = 100
noise_gauss = np.zeros(image_gray.shape, np.uint8)
cv2.randn(noise_gauss, mean, stddev)
```

```python
plt.imshow(noise_gauss, cmap="gray")
```

```python
# Salt and pepper
noise =  np.random.randint(0, 101, size = (image_gray.shape[0], image_gray.shape[1]), dtype=int)
zeros_pixel = np.where(noise == 0)
ones_pixel = np.where(noise == 100)
```

```python
bg_image = np.ones(image_gray.shape, np.uint8) * 128
```

```python
bg_image[zeros_pixel] = 0
bg_image[ones_pixel] = 255
```

```python
plt.imshow(bg_image, cmap="gray")
```

```python
image_noise_gauss = cv2.add(image_gray,noise_gauss)
```

```python
plt.imshow(image_noise_gauss, cmap="gray")
```

```python
from skimage.metrics import structural_similarity, mean_squared_error
mse_gauss = mean_squared_error(image_gray, image_noise_gauss)
(ssim, diff) = structural_similarity(image_gray, image_noise_gauss, full=True)
print(mse_gauss, ssim)
```

```python
image_gauss_median = cv2.medianBlur(image_noise_gauss, 3)
```

```python
mse_gauss_median = mean_squared_error(image_gray, image_gauss_median)
(ssim_gauss_median, diff) = structural_similarity(image_gray,  image_gauss_median, full=True)
```

```python
print(mse_gauss_median, ssim_gauss_median)
```

```python
plt.imshow(image_gauss_median, cmap="gray")
```

```python
import copy

image_sp = copy.deepcopy(image_gray)

image_sp[zeros_pixel] = 0
image_sp[ones_pixel] = 255
```

```python
plt.imshow(image_sp, cmap="gray")
```

```python
mse_sp = mean_squared_error(image_gray, image_sp)
(ssim_sp, diff) = structural_similarity(image_gray, image_sp, full=True)
print(mse_sp, ssim_sp)
```

```python
image_sp_median = cv2.medianBlur(image_sp, 3)
```

```python
mse_sp_median = mean_squared_error(image_gray, image_sp_median)
(ssim_sp_median, diff) = structural_similarity(image_gray, image_sp_median, full=True)
print(mse_sp_median, ssim_sp_median)
```

```python
import math

def geom(a):
    prod = 1
    for i in range(a.shape[0]):
        prod1 = 1
        for j in range(a.shape[1]):
            prod1 *= a[i,j]
        prod1 = math.pow(prod1, 1.0/9.0)
        prod *= prod1
    return prod

def proc(img, filter):
    img_res = copy.deepcopy(img)
    for i in range(0,img.shape[0] -2):
        for j in range(0,img.shape[1] -2):
            img_res[i:i+3, j:j+3] = filter(img[i:i+3, j:j+3])
    return img_res
    
res = proc(image_noise_gauss, geom)
```

```python
plt.imshow(res, cmap="gray")
```

```python
mse_geom = mean_squared_error(image_gray, res)
(ssim_geom, diff) = structural_similarity(image_gray, res, full=True)
print(mse_geom, ssim_geom)
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

image = cv2.imread('sar_1.jpg', cv2.IMREAD_GRAYSCALE)

plt.figure(figsize=(10, 6))
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.show()
```

```python
mean = 0
stddev = 30

noise_gauss = np.random.normal(mean, stddev, image.shape)

image_gauss = image.astype(np.float32) + noise_gauss
image_gauss = np.clip(image_gauss, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Исходное изображение')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(image_gauss, cmap='gray')
plt.title('Гауссовский шум')
plt.axis('off')

plt.show()
```

```python
a = -30
b = 30

noise_uniform = np.random.uniform(a, b, image.shape)

image_constant = image.astype(np.float32) + noise_uniform
image_constant = np.clip(image_constant, 0, 255).astype(np.uint8)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Исходное изображение')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(image_constant, cmap='gray')
plt.title('Постоянный шум')
plt.axis('off')

plt.show()
```

```python
median_3 = cv2.medianBlur(image_gauss, 3)
median_5 = cv2.medianBlur(image_gauss, 5)
median_7 = cv2.medianBlur(image_gauss, 7)

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image_gauss, cmap='gray')
plt.title('Гауссовский шум')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(median_3, cmap='gray')
plt.title('Медианный фильтр 3x3')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(median_5, cmap='gray')
plt.title('Медианный фильтр 5x5')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(median_7, cmap='gray')
plt.title('Медианный фильтр 7x7')
plt.axis('off')

plt.show()
```

```python
gauss_3 = cv2.GaussianBlur(image_gauss, (3, 3), 0)
gauss_5 = cv2.GaussianBlur(image_gauss, (5, 5), 0)
gauss_7 = cv2.GaussianBlur(image_gauss, (7, 7), 0)

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image_gauss, cmap='gray')
plt.title('Гауссовский шум')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gauss_3, cmap='gray')
plt.title('Фильтр Гаусса 3x3')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(gauss_5, cmap='gray')
plt.title('Фильтр Гаусса 5x5')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(gauss_7, cmap='gray')
plt.title('Фильтр Гаусса 7x7')
plt.axis('off')

plt.show()
```

```python
bilateral_1 = cv2.bilateralFilter(image_gauss, 5, 50, 50)
bilateral_2 = cv2.bilateralFilter(image_gauss, 9, 75, 75)
bilateral_3 = cv2.bilateralFilter(image_gauss, 15, 100, 100)

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image_gauss, cmap='gray')
plt.title('Гауссовский шум')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(bilateral_1, cmap='gray')
plt.title('Билатериальный фильтр: d=5')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(bilateral_2, cmap='gray')
plt.title('Билатериальный фильтр: d=9')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(bilateral_3, cmap='gray')
plt.title('Билатериальный фильтр: d=15')
plt.axis('off')

plt.show()
```

```python
nlm_1 = cv2.fastNlMeansDenoising(image_gauss, None, 5, 7, 21)
nlm_2 = cv2.fastNlMeansDenoising(image_gauss, None, 10, 7, 21)
nlm_3 = cv2.fastNlMeansDenoising(image_gauss, None, 20, 7, 21)

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(image_gauss, cmap='gray')
plt.title('Гауссовский шум')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(nlm_1, cmap='gray')
plt.title('Нелокальные средние: h=5')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(nlm_2, cmap='gray')
plt.title('Нелокальные средние: h=10')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(nlm_3, cmap='gray')
plt.title('Нелокальные средние: h=20')
plt.axis('off')

plt.show()
```
